```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, Output
from scipy import stats
from matplotlib.colors import hsv_to_rgb
from IPython.display import display, clear_output

# Настройки графики
plt.style.use('ggplot')
plt.rcParams['figure.dpi'] = 100

# Инициализация сетки
size = 0.02  # ±20 мм
points = 500
x = np.linspace(-size, size, points)
y = np.linspace(-size, size, points)
X, Y = np.meshgrid(x, y)
r = np.sqrt(X**2 + Y**2)

# Создаем отдельную область для вывода
output = Output()
display(output)

def calculate_intensity(R, lambda_center_nm, delta_lambda_nm, N_spectrum=100):
    """Расчет интенсивности"""
    lambda_center = lambda_center_nm * 1e-9
    delta_lambda = delta_lambda_nm * 1e-9
    
    d = r**2 / (2 * R)
    wavelengths = np.linspace(
        max(380e-9, lambda_center - delta_lambda/2),
        min(750e-9, lambda_center + delta_lambda/2),
        N_spectrum
    )
    
    total = np.zeros_like(r)
    for wl in wavelengths:
        phase = (4 * np.pi * d) / wl
        total += 2 * (1 + np.cos(phase))
    
    return total / np.max(total)

def wavelength_to_rgb(wavelength_nm, gamma=0.5):
    """Преобразование длины волны в RGB"""
    if 380 <= wavelength_nm <= 750:
        x = (wavelength_nm - 380)/(750 - 380)
        hue = 0.7*(1 - x)
        rgb = hsv_to_rgb([hue, 1.0, 1.0])
        return rgb ** gamma
    return np.zeros(3)

@interact(
    R=FloatSlider(value=15.0, min=1.0, max=20.0, step=0.1, description="R (м)"),
    lambda_center=FloatSlider(value=550, min=400, max=700, step=1, description="λ (нм)"),
    delta_lambda=FloatSlider(value=0, min=0, max=100, step=1, description="Δλ (нм)")
)
def update_rings(R, lambda_center, delta_lambda):
    with output:
        # Очищаем вывод без прокрутки
        clear_output(wait=True)
        
        # Закрываем все предыдущие фигуры
        plt.close('all')
        
        # Расчет данных
        if delta_lambda==0:
            delta_lambda=1e-9
        intensity = calculate_intensity(R, lambda_center, delta_lambda)
        
        # Генерация RGB
        wavelengths = np.linspace(
            max(380, lambda_center - delta_lambda//2),
            min(750, lambda_center + delta_lambda//2),
            100
        )
        rgb = np.zeros((*r.shape, 3))
        for wl in wavelengths:
            color = wavelength_to_rgb(wl).reshape(1, 1, 3)
            rgb += color * intensity[..., None]
        rgb /= np.max(rgb)
        
        # Усреднение профиля
        r_flat = r.ravel()
        I_flat = intensity.ravel()
        bins = np.linspace(0, r.max(), 300)
        radial_mean, _, _ = stats.binned_statistic(r_flat, I_flat, 
                                                 statistic='mean', bins=bins)
        r_centers = (bins[:-1] + bins[1:])/2 * 1000
        
        # Создаем фигуру
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))
        
        # Интерференционная картина
        ax1.imshow(rgb, 
                  extent=[-size*1000, size*1000, -size*1000, size*1000],
                  aspect='equal', 
                  origin='lower')
        ax1.set_title(f"R = {R:.1f} м\nλ = {lambda_center}±{delta_lambda} нм")
        ax1.grid(False)
        
        # Радиальный профиль
        ax2.plot(r_centers, radial_mean, 'b-', lw=2)
        ax2.grid(True)
        ax2.set_ylim(-0.1, 1.1)
        
        # Отображаем и сразу закрываем фигуру
        display(fig)
        plt.close(fig)

# Первый запуск
update_rings(15.0, 550, 0)
```

```python
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
import numpy as np
from scipy import stats

# Параметры для примеров
R = 10.0  # Радиус кривизны линзы

def generate_examples():
    """Генерация примеров для когерентного, квазикогерентного и белого света"""
    # Создаем фигуру
    fig = plt.figure(figsize=(15, 12))
    
    examples = [
        {'lambda_center': 550, 'delta_lambda': 0, 'title': 'Когерентный источник (λ=550 нм)', 'row': 0},
        {'lambda_center': 550, 'delta_lambda': 10, 'title': 'Квазикогерентный источник (λ=550±10 нм)', 'row': 1},
        {'lambda_center': 550, 'delta_lambda': 100, 'title': 'Белый свет (λ=550±100 нм)', 'row': 2}
    ]
    
    # Создаем подграфики (избегаем использования GridSpec для решения проблемы tight_layout)
    for i, example in enumerate(examples):
        # Расчет толщины зазора
        d = r**2 / (2 * R)
        
        # Монохроматический свет (специальная обработка)
        if example['delta_lambda'] == 0:
            lambda_m = example['lambda_center'] * 1e-9  # Перевод в метры
            phase = 4 * np.pi * d / lambda_m
            intensity = 2 * (1 + np.cos(phase))
            intensity = intensity / np.max(intensity)
            
            # Монохроматический свет отображаем в оттенках серого
            rgb = np.zeros((*r.shape, 3))
            for c in range(3):  # RGB каналы
                rgb[:,:,c] = intensity
        
        # Квазикогерентный и белый свет
        else:
            # Выбираем диапазон длин волн
            lambda_min = max(380, example['lambda_center'] - example['delta_lambda']//2)
            lambda_max = min(750, example['lambda_center'] + example['delta_lambda']//2)
            wavelengths = np.linspace(lambda_min, lambda_max, 50)
            
            # Создаем пустые массивы для накопления
            rgb = np.zeros((*r.shape, 3))
            total_intensity = np.zeros_like(r)
            
            # Для каждой длины волны
            for wl in wavelengths:
                # Расчет интенсивности
                lambda_m = wl * 1e-9  # Перевод в метры
                phase = 4 * np.pi * d / lambda_m
                intensity = 2 * (1 + np.cos(phase))
                intensity = intensity / np.max(intensity)
                
                # Добавляем к общей интенсивности
                total_intensity += intensity / len(wavelengths)
                
                # Преобразование длины волны в RGB
                color = wavelength_to_rgb(wl).reshape(1, 1, 3)
                
                # Добавляем взвешенный вклад к RGB изображению
                rgb += color * intensity[..., None] / len(wavelengths)
            
            # Нормализация RGB
            max_rgb = np.max(rgb)
            if max_rgb > 0:
                rgb = rgb / max_rgb
            
            # Для графика используем общую интенсивность
            intensity = total_intensity / np.max(total_intensity) if np.max(total_intensity) > 0 else total_intensity
        
        # Расчет радиального профиля
        r_flat = r.ravel()
        I_flat = intensity.ravel()
        bins = np.linspace(0, r.max(), 300)
        radial_mean, _, _ = stats.binned_statistic(r_flat, I_flat, 
                                                  statistic='mean', bins=bins)
        r_centers = (bins[:-1] + bins[1:])/2 * 1000  # Перевод в мм
        
        # Рисуем графики
        ax1 = plt.subplot(3, 2, i*2+1)
        ax2 = plt.subplot(3, 2, i*2+2)
        
        # Интерференционная картина
        ax1.imshow(rgb, extent=[-size*1000, size*1000, -size*1000, size*1000], 
                  aspect='equal', origin='lower')
        ax1.set_title(example['title'])
        ax1.set_xlabel('x, мм')
        ax1.set_ylabel('y, мм')
        ax1.grid(False)
        
        # Радиальный профиль
        ax2.plot(r_centers, radial_mean, 'r-', lw=2)
        ax2.set_title('Радиальный профиль интенсивности')
        ax2.set_xlabel('r, мм')
        ax2.set_ylabel('Относительная интенсивность')
        ax2.set_ylim(-0.1, 1.1)
        ax2.grid(True)
    
    # Используем tight_layout без дополнительных параметров для избежания проблем
    plt.suptitle('Сравнение интерференционных картин для различных источников света', fontsize=16, y=0.98)
    plt.tight_layout()
    plt.subplots_adjust(top=0.93)  # Регулируем верхнее поле для заголовка
    
    # Сохранение результата
    plt.savefig('newton_rings_comparison.png', dpi=300, bbox_inches='tight')
    return fig

# Генерация и отображение примеров
comparison_fig = generate_examples()
```
